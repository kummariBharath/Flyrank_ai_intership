## 1. My lane as an ML task

My lane is **Refresh / Content Opportunity Scoring**.

I frame this as a **ranking/scoring task** because the goal is not simply to predict whether a page is good or bad. The goal is to rank content pages by how much they deserve attention from a limited content-review team.

The output would be a **priority score** for each content page. Pages with higher scores would appear earlier in the review queue.

The decision this supports is: **Which pages should the content/SEO team review first?**

The eventual action could be to review a page for refresh, expansion, protection, pruning, or monitoring.

A ranking approach fits this problem because the team has limited review capacity. They do not need an automatic decision for every page; they need a useful shortlist of the pages that deserve attention first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## 2. Target or proxy

I will use **`is_declining_label`** as a proxy for pages showing a concerning performance outcome. It is an observed label that indicates whether a content page is currently classified as declining.

This proxy can help us identify pages that may deserve higher priority for human review. However, it does **not** prove that refreshing a page will cause its performance to recover. It represents observed decline, not the future causal effect of a content refresh.

For the eventual ranking, other signals such as impressions, sessions, CTR, content age, and search position may also help distinguish which pages deserve attention first.



## 3. Success metric

*One metric you can defend. What number means 'good'?*

## 3. Success metric

The primary success metric will be **Precision@K**.

Precision@K measures how many of the pages in the top K positions of the ranked review queue are relevant according to the observed proxy outcome. This matches the real decision because the content team has limited review capacity and needs the most useful pages to appear near the top of the queue.

For example, if the team can review 50 pages and 35 of the top 50 ranked pages are relevant according to the observed proxy, Precision@50 would be 70%.

The exact value of K should reflect the team's realistic review capacity rather than being chosen only because it produces a better-looking model result.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## 5. Why ML beats a fixed rule here

A simple fixed rule could prioritize pages using conditions such as high impressions combined with a declining trend. However, page-review priority may depend on several signals at the same time, including impressions, sessions, CTR, content age, search position, engagement, and content type.

Writing all of these relationships as hand-built rules could become complicated and difficult to maintain as the patterns change. A ranking model may be useful if these signals contain real but messy relationships that are difficult to express with a small set of fixed rules.

I cannot assume that ML will be better before testing it. Therefore, the eventual project should compare the ML ranking against a simple rule-based baseline and evaluate both using the same Precision@K metric.

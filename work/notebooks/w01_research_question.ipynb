Search question: Which pages should be reviewed first for a possible content refresh?

Decision: Decide which pages deserve priority in a limited content-review queue.

Action: A content/SEO reviewer could inspect the highest-ranked pages and decide whether to refresh, expand, protect, prune, or simply monitor them.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

path = "../../data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(path)

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nAverage 90-day impressions:",
      round(df["impressions_90d"].mean(), 2))

print("\nPages with declining trend:",
      (df["trend_direction"] == "down").sum())

print("\nPages with 500+ impressions:",
      (df["impressions_90d"] >= 500).sum())
```
